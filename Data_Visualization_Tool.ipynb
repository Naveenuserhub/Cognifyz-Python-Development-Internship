```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
data = pd.read_csv("D:\\files\\sales_data.csv")

# Display the dataset
print("Sales Data:")
print(data)

# Create a bar chart
plt.figure(figsize=(8, 5))
plt.bar(data["Month"], data["Sales"])

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=45)
plt.tight_layout()

# Display the chart
plt.show()
```
