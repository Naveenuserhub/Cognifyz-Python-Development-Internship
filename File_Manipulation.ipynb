```python
import os
import string

def count_words_in_file(filename):
    # 1. Check if file exists
    if not os.path.exists(filename):
        print(f" Error: The file '{filename}' does not exist.")
        return

    # 2. Read the file
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            text = file.read()
    except Exception as e:
        print(f" Error reading file: {e}")
        return

    # 3. Clean the text (remove punctuation and convert to lowercase)
    text = text.lower()
    translator = str.maketrans('', '', string.punctuation)
    cleaned_text = text.translate(translator)

    # 4. Split into words
    words = cleaned_text.split()

    # 5. Count word occurrences
    word_count = {}
    for word in words:
        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1

    # 6. Sort alphabetically
    sorted_words = sorted(word_count.items())

    # 7. Display results
    print(f"\n Word Count in '{filename}':")
    print("-" * 35)
    print(f"{'Word':<20} {'Count':>10}")
    print("-" * 35)
    for word, count in sorted_words:
        print(f"{word:<20} {count:>10}")
    print("-" * 35)
    print(f"Total unique words: {len(word_count)}")
    print(f"Total words: {len(words)}")

# --- STEP 1: Create a sample text file first ---
sample_text = """Python is great. Python is powerful and Python is easy to learn.
Data science with Python is amazing. Python python python!"""

with open("sample.txt", "w", encoding="utf-8") as f:
    f.write(sample_text)

print(" Sample file 'sample.txt' created successfully.")

# --- STEP 2: Run the word counter ---
count_words_in_file("sample.txt")
```
